# Multivariate Analysis: PCA, Clustering & MDS

This notebook applies PCA, clustering, and multidimensional scaling on real-world datasets.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.manifold import MDS
```

## Load Survey Data

```python
data = pd.read_csv("Survey.csv")
data.head()
```

## Data Preprocessing

```python
numeric_data = data.select_dtypes(include=['int64','float64'])

scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_data)
```

## PCA

```python
pca = PCA()
pca_result = pca.fit_transform(scaled_data)

plt.plot(pca.explained_variance_ratio_)
plt.title("PCA Variance Explained")
plt.xlabel("Components")
plt.ylabel("Variance")
plt.show()
```

## Clustering (K-Means)

```python
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(scaled_data)

data['Cluster'] = clusters
```

```python
plt.scatter(pca_result[:,0], pca_result[:,1], c=clusters)
plt.title("Cluster Visualization")
plt.show()
```

## MDS (Ice Cream Dataset)

```python
icecream = pd.read_csv("icecream.csv")

ice_numeric = icecream.select_dtypes(include=['int64','float64'])

scaler = StandardScaler()
ice_scaled = scaler.fit_transform(ice_numeric)
```

```python
mds = MDS(n_components=2, random_state=42)
mds_result = mds.fit_transform(ice_scaled)

plt.scatter(mds_result[:,0], mds_result[:,1])
plt.title("MDS Plot")
plt.show()
```

## Key Insights

- PCA reduces dimensionality and highlights key components
- Clustering identifies consumer segments
- MDS visualizes similarity between products
